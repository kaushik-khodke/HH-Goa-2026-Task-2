# Phase 6: Hybrid Retrieval & Rank Fusion

## Objective
Compare Dense Only vs. BM25 Only vs. Hybrid (BM25 + Dense RRF Fusion) to empirically verify if hybrid retrieval improves Recall@K and MRR@10.

```python
from src.retrieval.hybrid_retriever import ReciprocalRankFusion

rrf = ReciprocalRankFusion(k=60)
sparse_res = [({'passage_id': 'p1', 'text': 'Passage 1'}, 5.2), ({'passage_id': 'p2', 'text': 'Passage 2'}, 3.1)]
dense_res = [({'passage_id': 'p2', 'text': 'Passage 2'}, 0.88), ({'passage_id': 'p1', 'text': 'Passage 1'}, 0.75)]
fused = rrf.fuse(sparse_res, dense_res, top_k=2)
print("RRF Fused Rankings:", fused)
```

# Phase 5: Sparse Retrieval Benchmark (BM25)

## Objective
Evaluate BM25 sparse keyword search performance on Indic queries and passages using script-aware tokenization.

```python
from src.retrieval.bm25_retriever import BM25Retriever

retriever = BM25Retriever(k1=1.5, b=0.75)
corpus = [
    {"passage_id": "p1", "text": "मैनहट्टन परियोजना द्वितीय विश्व युद्ध का हिस्सा थी।"},
    {"passage_id": "p2", "text": "भारत एक विविध और सांस्कृतिक देश है।"}
]
retriever.index_corpus(corpus)
results = retriever.retrieve("मैनहट्टन परियोजना", top_k=2)
print("BM25 Search Results:", results)
```
